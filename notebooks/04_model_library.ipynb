## 1. Muat Data dan Preprocessing

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report, roc_curve, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns

URL = "https://raw.githubusercontent.com/ravi-arnan/machine-learning/main/data/healthcare-stroke-data.csv"
df = pd.read_csv(URL)

df = df[["age", "hypertension", "heart_disease", "avg_glucose_level", "bmi", "stroke"]].copy()
df["bmi"] = df["bmi"].fillna(df["bmi"].median())

print("Ukuran dataset:", df.shape)
print()
print("5 baris pertama:")
df.head()
```

```python
X = df.drop(columns=["stroke"]).values
y = df["stroke"].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f"Train: {X_train.shape[0]} baris")
print(f"Test:  {X_test.shape[0]} baris")

# Scaling untuk model yang sensitif skala (Logistic Regression, SVM, KNN)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nProporsi stroke:")
print(f"  Train: {y_train.mean():.3f}")
print(f"  Test:  {y_test.mean():.3f}")
```

## 2. Fungsi Uji Model

Untuk menghindari kode berulang.

```python
def evaluasi_model(model, X_test, y_test):
    """Prediksi dan cetak metrik."""
    y_pred = model.predict(X_test)

    hasil = {
        "akurasi": round(accuracy_score(y_test, y_pred), 4),
        "presisi": round(precision_score(y_test, y_pred, zero_division=0), 4),
        "recall": round(recall_score(y_test, y_pred, zero_division=0), 4),
        "f1": round(f1_score(y_test, y_pred, zero_division=0), 4),
    }
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    hasil["tp"] = tp
    hasil["tn"] = tn
    hasil["fp"] = fp
    hasil["fn"] = fn
    return hasil, y_pred


def cetak_hasil(nama, hasil):
    print(f"\n{'=' * 50}")
    print(f"  {nama}")
    print(f"{'=' * 50}")
    print(f"  Akurasi:  {hasil['akurasi']}")
    print(f"  Presisi:  {hasil['presisi']}")
    print(f"  Recall:   {hasil['recall']}")
    print(f"  F1-score: {hasil['f1']}")
    print(f"  TP={hasil['tp']}  FP={hasil['fp']}  FN={hasil['fn']}  TN={hasil['tn']}")
def plot_confusion_matrix(y_true, y_pred, judul, ax=None):
    """Visualisasi confusion matrix dalam bentuk heatmap."""
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        ax=ax,
        xticklabels=["Pred: Tidak (0)", "Pred: Stroke (1)"],
        yticklabels=["Aktual: Tidak (0)", "Aktual: Stroke (1)"],
    )
    if ax:
        ax.set_title(judul, fontsize=10)
        ax.set_xlabel("Prediksi Label")
        ax.set_ylabel("Aktual Label")
    else:
        plt.title(judul, fontsize=11)
        plt.xlabel("Prediksi Label")
        plt.ylabel("Aktual Label")
        plt.tight_layout()
        plt.show()
```

## 3. Logistic Regression

```python
# Tanpa scaling vs dengan scaling
for nama, X_tr, X_te in [
    ("Logistic Regression (tanpa scaling)", X_train, X_test),
    ("Logistic Regression (dengan scaling)", X_train_scaled, X_test_scaled),
]:
    for imbang in [None, "balanced"]:
        label = f"{nama}, class_weight={imbang}"
        model = LogisticRegression(max_iter=5000, class_weight=imbang)
        model.fit(X_tr, y_train)
        hasil, _ = evaluasi_model(model, X_te, y_test)
        cetak_hasil(label, hasil)
```

## 4. Decision Tree

```python
for kedalaman in [3, 5, 10, None]:
    for imbang in [None, "balanced"]:
        label = f"Decision Tree (max_depth={kedalaman}, class_weight={imbang})"
        model = DecisionTreeClassifier(max_depth=kedalaman, class_weight=imbang, random_state=42)
        model.fit(X_train, y_train)
        hasil, _ = evaluasi_model(model, X_test, y_test)
        cetak_hasil(label, hasil)
```

## 5. Random Forest

```python
for n in [50, 100, 200]:
    for imbang in [None, "balanced"]:
        label = f"Random Forest (n={n}, class_weight={imbang})"
        model = RandomForestClassifier(n_estimators=n, class_weight=imbang, random_state=42)
        model.fit(X_train, y_train)
        hasil, _ = evaluasi_model(model, X_test, y_test)
        cetak_hasil(label, hasil)
```

## 6. SVM (Support Vector Machine)

```python
for kernel in ["linear", "rbf"]:
    for imbang in [None, "balanced"]:
        label = f"SVM ({kernel}, class_weight={imbang})"
        model = SVC(kernel=kernel, class_weight=imbang, random_state=42)
        model.fit(X_train_scaled, y_train)
        hasil, _ = evaluasi_model(model, X_test_scaled, y_test)
        cetak_hasil(label, hasil)
```

## 7. Perbandingan Semua Model (F1-score)

Rangkuman model terbaik tiap keluarga.

```python
model_terbaik = [
    ("Logistic Regression + scaling + balanced",
     LogisticRegression(max_iter=5000, class_weight="balanced").fit(X_train_scaled, y_train),
     X_test_scaled),
    ("Decision Tree (max_depth=5, balanced)",
     DecisionTreeClassifier(max_depth=5, class_weight="balanced", random_state=42).fit(X_train, y_train),
     X_test),
    ("Random Forest (n=100, balanced)",
     RandomForestClassifier(n_estimators=100, class_weight="balanced", random_state=42).fit(X_train, y_train),
     X_test),
    ("SVM (rbf, balanced)",
     SVC(kernel="rbf", class_weight="balanced", random_state=42).fit(X_train_scaled, y_train),
     X_test_scaled),
]

print(f"{'Model':<45} {'Akurasi':>8} {'Presisi':>8} {'Recall':>8} {'F1':>8}")
print("-" * 77)

for nama, model, X_te in model_terbaik:
    hasil, _ = evaluasi_model(model, X_te, y_test)
    print(f"{nama:<45} {hasil['akurasi']:>8} {hasil['presisi']:>8} {hasil['recall']:>8} {hasil['f1']:>8}")
```

### 7.1 Visualisasi Confusion Matrix Model Terbaik

Perbandingan visual matriks konfusi (True Negative, False Positive, False Negative, True Positive) untuk keempat model klasifikasi terbaik:

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 8.5))
axes = axes.ravel()

for i, (nama, model, X_te) in enumerate(model_terbaik):
    hasil, y_pred = evaluasi_model(model, X_te, y_test)
    rec, prec, f1 = hasil["recall"], hasil["presisi"], hasil["f1"]
    judul = f"{nama}\nRecall: {rec:.2f} | Presisi: {prec:.2f} | F1: {f1:.2f}"
    plot_confusion_matrix(y_test, y_pred, judul, ax=axes[i])

plt.suptitle("Perbandingan Confusion Matrix Model Terbaik Tiap Algoritma", fontsize=13, y=0.99)
plt.tight_layout()
plt.show()
```

### 7.2 Perbandingan Metrik Evaluasi (Grouped Bar Chart)

Visualisasi perbandingan Akurasi, Presisi, Recall, dan F1-Score dari keempat model terbaik. Terlihat jelas fenomena *accuracy paradox*: akurasi tinggi tidak menjamin model mampu mendeteksi pasien berisiko stroke (Recall tinggi).

```python
data_metrik = []
for nama, model, X_te in model_terbaik:
    hasil, _ = evaluasi_model(model, X_te, y_test)
    data_metrik.append({
        "Model": nama.split("(")[0].replace("+ scaling + balanced", "").strip(),
        "Akurasi": hasil["akurasi"],
        "Presisi": hasil["presisi"],
        "Recall": hasil["recall"],
        "F1-Score": hasil["f1"]
    })

df_plot = pd.DataFrame(data_metrik).set_index("Model")
ax = df_plot.plot(kind="bar", figsize=(10, 5), colormap="viridis", edgecolor="black", width=0.75)
plt.title("Perbandingan Metrik Utama 4 Model (class_weight=balanced)", fontsize=13, pad=12)
plt.ylabel("Skor", fontsize=11)
plt.xlabel("Model Klasifikasi", fontsize=11)
plt.xticks(rotation=0)
plt.ylim(0, 1.05)
plt.legend(loc="upper right", framealpha=0.9)
plt.grid(axis="y", linestyle="--", alpha=0.4)

for p in ax.patches:
    val = p.get_height()
    if val > 0.05:
        ax.annotate(f"{val:.2f}", (p.get_x() + p.get_width() / 2., val / 2),
                    ha="center", va="center", color="white", fontsize=8, fontweight="bold", rotation=90)

plt.tight_layout()
plt.show()
```

### 7.3 Perbandingan Kurva ROC dan Skor ROC-AUC

Kurva ROC (Receiver Operating Characteristic) mengukur kemampuan diskriminasi model pada berbagai ambang probabilitas tanpa terpengaruh oleh rasio ketidakseimbangan kelas.

```python
plt.figure(figsize=(8.5, 6))

for nama, model, X_te in model_terbaik:
    score = model.predict_proba(X_te)[:, 1] if hasattr(model, "predict_proba") else model.decision_function(X_te)
    fpr, tpr, _ = roc_curve(y_test, score)
    auc = roc_auc_score(y_test, score)
    label_model = nama.split("(")[0].replace("+ scaling + balanced", "").strip()
    plt.plot(fpr, tpr, lw=2, label=f"{label_model} (AUC = {auc:.3f})")

plt.plot([0, 1], [0, 1], color="grey", lw=1.5, linestyle="--", label="Random Guess (AUC = 0.500)")
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.02])
plt.xlabel("False Positive Rate (1 - Spesifisitas)", fontsize=11)
plt.ylabel("True Positive Rate (Recall / Sensitivitas)", fontsize=11)
plt.title("Kurva ROC 4 Model Klasifikasi Stroke", fontsize=13, pad=12)
plt.legend(loc="lower right", fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 7.4 Peringkat Kepentingan Fitur (Feature Importance)

Menjawab pertanyaan klinis: faktor apa yang paling berkontribusi terhadap risiko stroke berdasarkan model linier (Logistic Regression) dan pohon keputusan (Random Forest)?

```python
fitur_nama = ["age", "hypertension", "heart_disease", "avg_glucose_level", "bmi"]
fitur_label = ["Usia (age)", "Hipertensi", "Penyakit Jantung", "Gula Darah (glucose)", "Indeks Massa Tubuh (BMI)"]

m_lr = model_terbaik[0][1]
m_rf = model_terbaik[2][1]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8))

# Koefisien Logistic Regression
s_lr = pd.Series(m_lr.coef_[0], index=fitur_label).sort_values()
s_lr.plot(kind="barh", ax=ax1, color="#1f77b4", edgecolor="black")
ax1.set_title("Koefisien Fitur (Logistic Regression)\n(Arah dan Bobot Pengaruh)", fontsize=11)
ax1.set_xlabel("Nilai Koefisien (Log-Odds)", fontsize=10)
ax1.grid(axis="x", linestyle="--", alpha=0.4)

# Feature Importance Random Forest
s_rf = pd.Series(m_rf.feature_importances_, index=fitur_label).sort_values()
s_rf.plot(kind="barh", ax=ax2, color="#2ca02c", edgecolor="black")
ax2.set_title("Feature Importance (Random Forest)\n(Tingkat Kontribusi Total)", fontsize=11)
ax2.set_xlabel("Kepentingan Relatif (Gini Importance)", fontsize=10)
ax2.grid(axis="x", linestyle="--", alpha=0.4)

plt.suptitle("Analisis Faktor Risiko Stroke dari Model Library", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

### 7.5 Visualisasi Alur Keputusan (Decision Tree Rules)

Diagram pohon keputusan yang mengilustrasikan logika pembagian (*split*) data secara intuitif. Sangat cocok disajikan pada slide untuk menjelaskan bagaimana model mengambil keputusan klinis bertahap.

```python
m_dt = model_terbaik[1][1]

plt.figure(figsize=(15, 7.5))
plot_tree(
    m_dt,
    max_depth=2,
    feature_names=["age", "hypertension", "heart_disease", "avg_glucose_level", "bmi"],
    class_names=["Tidak Stroke", "Risiko Stroke"],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Alur Keputusan Klinis Model Decision Tree (Kedalaman 2)", fontsize=13, pad=12)
plt.tight_layout()
plt.show()
```
